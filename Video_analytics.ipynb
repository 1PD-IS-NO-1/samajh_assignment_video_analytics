# **Fashion Image Classifier**

```python
# Simple Fashion Test for Instagram Image
# Tests only: https://www.instagram.com/p/DJSptl7Sr5q

!pip install tensorflow pillow requests numpy

import requests
import numpy as np
from PIL import Image
from io import BytesIO
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.efficientnet import preprocess_input, decode_predictions

# Load model
print("Loading model...")
try:
    model = EfficientNetB0(weights='imagenet')
    print("✅ Model loaded")
except Exception as e:
    print(f"❌ Model failed: {e}")
    model = None

# Fashion keywords
fashion_keywords = [
    'dress', 'shirt', 'jean', 'suit', 'jacket', 'coat', 'vest', 'sweater',
    'cardigan', 'blazer', 'trouser', 'pants', 'skirt', 'shorts', 'shoe',
    'boot', 'sneaker', 'sandal', 'hat', 'cap', 'scarf', 'tie', 'belt',
    'watch', 'bag', 'purse', 'bikini', 'swimsuit', 'underwear', 'bra',
    'sunglass', 'uniform', 'garment', 'apparel', 'clothing', 'wear'
]

def test_instagram_fashion(instagram_url):
    """Simple test: Fashion or Not Fashion"""
    print(f"🔍 Testing: {instagram_url}")

    # Convert Instagram URL to direct image URL
    direct_url = instagram_url.rstrip('/') + '/media/?size=l'
    print(f"📸 Direct URL: {direct_url}")

    try:
        # Download image
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
        }
        response = requests.get(direct_url, headers=headers, timeout=30)
        response.raise_for_status()

        # Load image
        img = Image.open(BytesIO(response.content)).convert("RGB")
        print(f"✅ Image loaded: {img.size}")

        # Preprocess for model
        img_resized = img.resize((224, 224))
        img_array = image.img_to_array(img_resized)
        img_array = np.expand_dims(img_array, axis=0)
        img_array = preprocess_input(img_array)

        # Get predictions
        preds = model.predict(img_array, verbose=0)
        decoded_preds = decode_predictions(preds, top=10)[0]

        fashion_score = 0.0
        fashion_items = []

        for _, class_name, confidence in decoded_preds:
            class_lower = class_name.lower()
            for keyword in fashion_keywords:
                if keyword in class_lower:
                    fashion_score += confidence
                    fashion_items.append(f"{class_name} ({confidence:.3f})")
                    break

        # Decision
        is_fashion = fashion_score >= 0.25

        # Results
        print("\n" + "="*50)
        print("🎯 RESULT:")
        print(f"{'✅ FASHION IMAGE' if is_fashion else '❌ NOT FASHION IMAGE'}")
        print(f"Fashion Score: {fashion_score:.3f}")
        print(f"Threshold: 0.25")

        if fashion_items:
            print(f"\nDetected Fashion Items:")
            for item in fashion_items[:3]:
                print(f"  • {item}")

        print(f"\nTop 5 Predictions:")
        for i, (_, class_name, conf) in enumerate(decoded_preds[:5], 1):
            print(f"  {i}. {class_name}: {conf:.3f}")

        return is_fashion, fashion_score

    except Exception as e:
        print(f"❌ Error: {e}")
        print("\n💡 If this fails, try:")
        print("1. Save the Instagram image manually")
        print("2. Upload to imgur.com")
        print("3. Use that URL instead")
        return None, None

# Test your specific Instagram URL
if __name__ == "__main__":
    print("🚀 INSTAGRAM FASHION TEST")
    print("="*50)

    # Your Instagram URL
    url = "https://www.instagram.com/p/DJSptl7Sr5q"

    if model is not None:
        result = test_instagram_fashion(url)
    else:
        print("❌ Cannot test - model not loaded")
```

# **YOLO12L to detect person in Video (Max Detection Accuracy)**

## **YOLLO12 MODEL SIZE**

* **S- SMALL**
* **M- Medium**
* **l- Large**
* **x- Extra_large**



```python
!pip install -q tensorflow tensorflow-hub opencv-python numpy ultralytics onnxruntime
!pip install -q torch torchvision deep-sort-realtime
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
import os
import time

# Load the YOLOv12 model
model = YOLO("/content/yolo12l.pt")

PERSON_CLASS_ID = 0
start_time = time.time()
def process_frame(frame):

    results = model(frame, conf=0.3, iou=0.6, imgsz=1280, half=True, augment=False)[0]
    annotated_frame = frame.copy()

    # Process detections
    for det in results.boxes:
        if det.cls.item() == PERSON_CLASS_ID:  # Filter for "person" class
            # Get bounding box coordinates
            x1, y1, x2, y2 = map(int, det.xyxy[0])
            confidence = det.conf.item()


            cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            label = f"Person: {confidence:.2f}"
            cv2.putText(annotated_frame, label, (x1, y1 - 10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    return annotated_frame, results.speed

def main(input_video_path, output_video_path):


    cap = cv2.VideoCapture(input_video_path)

    if not cap.isOpened():
        print(f"Error: Could not open video file {input_video_path}")
        return

    # Get video properties
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # Initialize video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec for MP4
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

    if not out.isOpened():
        print(f"Error: Could not create output video file {output_video_path}")
        cap.release()
        return

    # Track inference speed
    inference_times = []
    total_start_time = time.time()

    try:
        frame_count = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                print("End of video or error reading frame.")
                break

            # Process the frame for human detection
            annotated_frame, speed = process_frame(frame)

            # Write the annotated frame to the output video
            out.write(annotated_frame)

            # Record inference time
            inference_times.append(speed['inference'])
            frame_count += 1

            # Print progress
            print(f"Processed frame {frame_count}/{total_frames}, Inference time: {speed['inference']:.2f}ms")

    finally:
        # Release resources
        cap.release()
        out.release()


        total_time = time.time() - total_start_time
        avg_inference_time = np.mean(inference_times) if inference_times else 0
        avg_fps = 1000 / avg_inference_time if avg_inference_time > 0 else 0
        print(f"\nProcessing Summary:")
        print(f"Total frames processed: {frame_count}")
        print(f"Total processing time: {total_time:.2f} seconds")
        print(f"Average inference time per frame: {avg_inference_time:.2f} ms")
        print(f"Average FPS: {avg_fps:.2f}")

if __name__ == "__main__":

    input_video = "/content/trimmed_video.mp4"
    output_video = "output_video.mp4"
    main(input_video, output_video)
print(time.time()- start_time)
```

# **YOLO12S Model To Detect person in Video**

```python
import cv2
from ultralytics import YOLO
import numpy as np
import os
import time
start_time = time.time()
# Load the YOLOv12 model
model = YOLO("/content/yolo12s.pt")  # Use medium model for better accuracy; can use yolov12l.pt for even higher accuracy

# Define the class ID for "person" (COCO dataset class ID 0)
PERSON_CLASS_ID = 0

def process_frame(frame):
    """
    Process a single frame for human detection using YOLOv12.
    Args:
        frame: Input image or video frame
    Returns:
        Annotated frame with bounding boxes around detected humans
    """
    # Perform inference with TTA and FP16 for speed
    results = model(frame, conf=0.15, iou=0.7, imgsz=640, half=True, augment=True)[0]  # Lower conf, TTA enabled

    # Get the annotated frame
    annotated_frame = frame.copy()

    # Process detections
    for det in results.boxes:
        if det.cls.item() == PERSON_CLASS_ID:  # Filter for "person" class
            # Get bounding box coordinates
            x1, y1, x2, y2 = map(int, det.xyxy[0])
            confidence = det.conf.item()

            # Draw bounding box and label
            cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            label = f"Person: {confidence:.2f}"
            cv2.putText(annotated_frame, label, (x1, y1 - 10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    return annotated_frame, results.speed  # Return speed metrics

def main(input_video_path, output_video_path):
    """
    Main function to process an input video file and save the output with human detections.
    Args:
        input_video_path: Path to the input video file
        output_video_path: Path to save the output video file
    """
    # Initialize video capture
    cap = cv2.VideoCapture(input_video_path)

    if not cap.isOpened():
        print(f"Error: Could not open video file {input_video_path}")
        return

    # Get video properties
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # Initialize video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec for MP4
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

    if not out.isOpened():
        print(f"Error: Could not create output video file {output_video_path}")
        cap.release()
        return

    # Track inference speed
    inference_times = []
    total_start_time = time.time()

    try:
        frame_count = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                print("End of video or error reading frame.")
                break

            # Process the frame for human detection
            annotated_frame, speed = process_frame(frame)

            # Write the annotated frame to the output video
            out.write(annotated_frame)

            # Record inference time
            inference_times.append(speed['inference'])
            frame_count += 1

            # Print progress
            print(f"Processed frame {frame_count}/{total_frames}, Inference time: {speed['inference']:.2f}ms")

    finally:
        # Release resources
        cap.release()
        out.release()

        # Calculate and print speed metrics
        total_time = time.time() - total_start_time
        avg_inference_time = np.mean(inference_times) if inference_times else 0
        avg_fps = 1000 / avg_inference_time if avg_inference_time > 0 else 0
        print(f"\nProcessing Summary:")
        print(f"Total frames processed: {frame_count}")
        print(f"Total processing time: {total_time:.2f} seconds")
        print(f"Average inference time per frame: {avg_inference_time:.2f} ms")
        print(f"Average FPS: {avg_fps:.2f}")

if __name__ == "__main__":
    # Specify input and output video paths
    input_video = "/content/trimmed_video.mp4"  # Replace with your input video path
    output_video = "output_video.mp4"  # Output video will be saved here
    main(input_video, output_video)

print(time.time() - start_time)
```
